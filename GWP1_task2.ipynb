    return beta0 + beta1 * ((1 - np.exp(-lamb * t)) / (lamb * t)) + beta2 * (((1 - np.exp(-lamb * t)) / (lamb * t)) - np.exp(-lamb * t))

```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

# Hypothetical Danish data
maturities = np.array([0.5, 1, 2, 3, 5, 7, 10, 15, 20, 30])
yields = np.array([-0.60, -0.55, -0.50, -0.40, -0.20, 0.05, 0.25, 0.50, 0.70, 0.90])

def nelson_siegel(t, beta0, beta1, beta2, lamb):
    return beta0 + beta1 * ((1 - np.exp(-lamb * t)) / (lamb * t)) + beta2 * (((1 - np.exp(-lamb * t)) / (lamb * t)) - np.exp(-lamb * t))

# Initial guess for parameters
p0 = [0.5, -1, -0.5, 0.5]

# Fit the model
popt, pcov = curve_fit(nelson_siegel, maturities, yields, p0=p0, maxfev=5000)

# Fitted parameters
beta0_ns, beta1_ns, beta2_ns, lamb_ns = popt

# Predicted yields from Nelson-Siegel
yields_ns_fitted = nelson_siegel(maturities, beta0_ns, beta1_ns, beta2_ns, lamb_ns)
rmse_ns = np.sqrt(mean_squared_error(yields, yields_ns_fitted))

# Plot the results
plt.plot(maturities, yields, 'o', label='Actual Yields (Denmark)')
plt.plot(maturities, yields_ns_fitted, '-', label=f'Nelson-Siegel Fit (RMSE: {rmse_ns:.4f})')
plt.xlabel('Maturity (Years)')
plt.ylabel('Yield (%)')
plt.title('Nelson-Siegel Model Fit for Danish Government Bonds')
plt.legend()
plt.grid(True)
plt.show()

print(f"Nelson-Siegel Parameters:")
print(f"Beta0 (Long-term level): {beta0_ns:.4f}")
print(f"Beta1 (Short-term level): {beta1_ns:.4f}")
print(f"Beta2 (Medium-term curvature): {beta2_ns:.4f}")
print(f"Lambda (Decay factor): {lamb_ns:.4f}")
```

d. Fit a Cubic-Spline Model:

```python
from scipy.interpolate import CubicSpline

# Fit the cubic spline
cs = CubicSpline(maturities, yields)

# Predicted yields from Cubic Spline
yields_cs_fitted = cs(maturities)
rmse_cs = np.sqrt(mean_squared_error(yields, yields_cs_fitted))

# Plot the results
plt.plot(maturities, yields, 'o', label='Actual Yields (Denmark)')
plt.plot(maturities, yields_cs_fitted, '-', label=f'Cubic Spline Fit (RMSE: {rmse_cs:.4f})')
plt.xlabel('Maturity (Years)')
plt.ylabel('Yield (%)')
plt.title('Cubic Spline Model Fit for Danish Government Bonds')
plt.legend()
plt.grid(True)
plt.show()

print(f"Cubic Spline RMSE: {rmse_cs:.4f}")
```

By visually inspecting the plots and comparing the Root Mean Squared Error (RMSE) values, we can assess the fit. Generally:

Ethics of Smoothing Data (Nelson-Siegel):

In Module 2 Lesson 4, we discussed that smoothing data can be unethical if it's done to misrepresent the underlying reality or to support a pre-determined conclusion without justification.

However, the smoothing performed by the Nelson-Siegel model in the context of yield curve fitting is generally not considered unethical for the following reasons:

Purpose is to Extract Underlying Structure: The goal of fitting a Nelson-Siegel model is to capture the fundamental economic factors (level, slope, curvature) that drive the yield curve, rather than focusing on every minor fluctuation which might be due to temporary market noise or illiquidity in specific maturities.
Economic Interpretation: The model provides parameters with economic meaning, allowing for analysis of market expectations about future interest rates and economic conditions.
Common Practice: Yield curve modeling is a standard practice in finance for valuation, risk management, and forecasting. Parametric models like Nelson-Siegel are widely used and accepted.
Transparency: The model and its parameters are explicitly stated, allowing others to understand how the smoothing is being applied.
Not Hiding Information: The smoothing doesn't erase the actual data points. The model is fitted to these points, and the degree of fit (e.g., RMSE) can be evaluated. Significant deviations between the model and the actual data would raise concerns about the model's adequacy, not necessarily its ethical use.
