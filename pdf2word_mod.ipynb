```python
import os
from pdf2docx import Converter
import tkinter as tk
from tkinter import filedialog, messagebox

def pdf_to_word(pdf_path, output_dir):
    if not os.path.isfile(pdf_path):
        raise FileNotFoundError(f"PDF file not found: {pdf_path}")
    if not os.path.isdir(output_dir):
        os.makedirs(output_dir, exist_ok=True)
    base_name = os.path.splitext(os.path.basename(pdf_path))[0]
    docx_file = os.path.join(output_dir, f"{base_name}.docx")
    cv = Converter(pdf_path)
    cv.convert(docx_file, start=0, end=None)
    cv.close()
    return docx_file

def select_pdf():
    file_path = filedialog.askopenfilename(
        title="Select PDF file",
        filetypes=[("PDF files", "*.pdf")]
    )
    pdf_entry.delete(0, tk.END)
    pdf_entry.insert(0, file_path)

def select_output_dir():
    dir_path = filedialog.askdirectory(title="Select Output Directory")
    output_entry.delete(0, tk.END)
    output_entry.insert(0, dir_path)

def convert():
    pdf_path = pdf_entry.get().strip()
    output_dir = output_entry.get().strip()
    try:
        docx_file = pdf_to_word(pdf_path, output_dir)
        messagebox.showinfo("Success", f"Conversion successful!\nDOCX file saved at:\n{docx_file}")
    except Exception as e:
        messagebox.showerror("Error", str(e))

if __name__ == "__main__":
    root = tk.Tk()
    root.title("PDF to Word Converter")

    tk.Label(root, text="PDF File:").grid(row=0, column=0, padx=5, pady=5, sticky="e")
    pdf_entry = tk.Entry(root, width=40)
    pdf_entry.grid(row=0, column=1, padx=5, pady=5)
    tk.Button(root, text="Browse", command=select_pdf).grid(row=0, column=2, padx=5, pady=5)

    tk.Label(root, text="Output Directory:").grid(row=1, column=0, padx=5, pady=5, sticky="e")
    output_entry = tk.Entry(root, width=40)
    output_entry.grid(row=1, column=1, padx=5, pady=5)
    tk.Button(root, text="Browse", command=select_output_dir).grid(row=1, column=2, padx=5, pady=5)

    tk.Button(root, text="Convert", command=convert, width=20).grid(row=2, column=0, columnspan=3, pady=15)

    root.mainloop()
```
